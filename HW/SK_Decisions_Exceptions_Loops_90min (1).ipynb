## 1. Nested decision: library access (10 minutes)

Write a function named `check_library_access()` with no arguments and no return. Ask for a student's age and whether they have a library card (`yes` or `no`). Use a **nested** `if` structure to print one appropriate message:

- Age 18 or older with a card: may borrow books.
- Age 18 or older without a card: may apply for a card.
- Under 18 with a card: may borrow with a guardian's permission.
- Under 18 without a card: must bring a guardian to apply.

Call the function below.

```python
# Write your function here
def check_library_access(): # Defines a function
  try:
    # Asks users for input
    age = int(input("Enter your age: "))
    card = input("Do you have a card? yes/no").lower()
    # A nested if/else statement
    if age >= 18:
      if card == "yes":
        print("May borrow books.")
      else:
        print("May apply for a card.")
    # A nested if/else statement
    else:
      if card == "yes":
        print("May borrow with a guardian's permission.")
      else:
        print("Must bring a guardian to apply.")
  # ValueError occurs if integer isn't entered
  except ValueError:
    print("Please enter an integer only.")

check_library_access() # Calls the function
```

## 2. Flattened decision with compound logic (10 minutes)

Write a function named `recommend_activity()` with no arguments and no return. Ask for temperature as an integer and whether it is raining (`yes` or `no`). Use a **flattened** `if / elif / else` structure with compound logic:

- If it is raining **or** the temperature is below 45: recommend an indoor activity.
- Otherwise, if the temperature is from 45 through 75 inclusive: recommend a walk.
- Otherwise: recommend an outdoor game.

Call the function.

```python
# Write your function here
def recommend_activity(): # Defines a function
  # User input with try/except
  try:

    temp = int(input("What temperature is it? "))
    raining = input("Is it raining? yes/no: ").lower()
    # Flattened if/elif/else statement
    if temp < 45 or raining == "yes":
      print("Play a board game!")
    elif temp >= 45 and temp <= 75:
      print("Go take a walk.")
    else:
      print("Go play a game of catch!")

  # ValueError if incorrect input entered
  except ValueError:
    print("Please enter an integer.")

recommend_activity() # Calls the function
```

## 4. Raise a meaningful error (12 minutes)

Write `get_course_level()` with no arguments and no return. Ask for a course level as an integer.

- Use `try` to convert the input.
- If the number is not from 100 through 499 inclusive, use `raise ValueError(...)` with a useful message.
- Use `except ValueError as e` to print `Invalid course level:` followed by the error message.
- Use `else` to print a confirmation only when the value is valid.
- Use `finally` to print `Course-level check complete.` every time.

Test it with `abc`, `50`, and `141`.

```python
# Write your function here
def get_course_level(): # Defines the function
  # Try/except for user input
  try:
    course_level = int(input("Enter your course level: "))
    # Flattened if/else
    if course_level < 100 or course_level > 499:
      raise ValueError("Please enter a valid course level.")
    else:
      print("Confirmed course level.")
  # ValueError if incorrect input entered
  except ValueError as e:
    print("Invalid course level: Please enter a valid course level.")
  # Prints the following after every outcome of the if/else statement
  finally:
    print("Course-level check complete.")

get_course_level() # Calls the function
```

## 6. `range` and a running total (12 minutes)

Write `sum_even_numbers()` with no arguments and no return. Ask the user for a positive integer `n`. Make certain to validate your input using try..except with raise.  Use `for number in range` to calculate and print the sum of all even numbers from 2 through `n`.

Example: if `n` is 8, the total is `2 + 4 + 6 + 8`, or 20.


```python
# Write your function here
def sum_even_numbers(): # Defines the function
  # Try/except for user input
  try:
    n = int(input("Enter a positive integer: "))
    # If statement with a value error if negative integer entered
    if n <= 0:
      raise ValueError("Please enter a positive integer.")
    # Total value is 0
    total = 0
    # For every number in the range, do the following operation and then save the answer in total
    for number in range(2, n+1, 2):
      total = total + number
    # Print the final sum
    print("The sum is: ", total)
  # Value error if negative integer entered
  except ValueError:
    print("Please enter a positive integer.")

sum_even_numbers() # Calls the function
```

## 7. `enumerate` (10 minutes)

Use the provided list. Write `number_topics()` with no arguments and no return. Use `enumerate(topics, start=1)` to print output in this format:

`1. variables`

Do not calculate positions using `topics.index(...)`.

Note : the start= tells enumerate what number to start counting from, the default is zero.  


```python
topics = ["variables", "decisions", "loops", "functions"]

def number_topics(): # Defines the function
  # Enumerate and print every item in the list, starting with the number 1
  for number, topic in enumerate(topics, start=1):
    print(str(number) + ". " + topic)

number_topics() # Calls the function
```

## 8. `continue` and `break` (13 minutes)

Write `collect_scores()` with no arguments and no return. Repeatedly ask for a whole-number score.

- If the user types `done`, end the loop with `break`.
- Convert other entries to integers using `try`.
- If an entry is not an integer, print a message and use `continue`.
- If a score is outside 0–100, print a message and use `continue`.
- For each valid score, add it to `total` and add 1 to `valid_scores`.
- After the loop, print the average if at least one valid score was entered; otherwise print a suitable message.

**Hint:** Because you must check for `done` before conversion, first store the input as text using `.strip().lower()`.


```python
def collect_scores(): # Defines the function
  # Set values of total and valid_scores
  total = 0
  valid_scores = 0

  # While the value is true, ask user for a score or if done- creates a loop
  while True:

    user_score = input("Enter a score or type done: ").strip().lower()
    # If the user is done, break the loop
    if user_score == "done":
      break
    # Try/except for input value
    try:
      score = int(user_score)

    except ValueError:
      print("Please enter a whole number.")

    # If the score is not between 0-100, print the following message and continue the loop
    if score < 0 or score > 100:
      print("Please enter a number between 0-100")
      continue
    # New values for total and valid_scores
    total = total + score
    valid_scores = valid_scores + 1

  # If/else statement to print the average score
  if valid_scores > 0:
      average = total / valid_scores
      print("The average is:", average)

  else:
      print("No valid scores were entered.")


collect_scores() # Calls the function

```
