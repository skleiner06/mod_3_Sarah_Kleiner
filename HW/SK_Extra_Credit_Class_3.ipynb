
## 2) Triangle validity & type

**Prompt**  
Ask for three side lengths **a**, **b**, **c** (floats).  
- First, validate that all are **> 0**.  
- Then check triangle inequality: `a + b > c`, `a + c > b`, `b + c > a`.  
  - If invalid → print `"Not a triangle"` and stop.  
- If valid, print **exactly one** of: `"Equilateral"`, `"Isosceles"`, `"Scalene"`.

**Your task**  
- Validate numeric conversion and positivity.  
- Implement the checks carefully and choose exactly one type to print.

*Gentle hint:* Check validity first; only classify if valid.

**Support for triangles **

A triangle is Equilateral if All three sides are equal:
a == b == c

Isosceles : Exactly two sides are equal:
(a == b or a == c or b == c) but not all three equal

Scalene : All three sides are different:
a != b and a != c and b != c




```python
# Defines the function
def triangle_check():

  # Try/except for user input
  try:
    a = float(input("Enter a value for side a: "))
    b = float(input("Enter a value for side b: "))
    c = float(input("Enter a value for side c: "))

    # If any of the sides are less than or equal to 0, user is told that it is not a triangle and is prompted to re-enter values
    if a <= 0 or b <= 0 or c <= 0:
      print("Not a triangle.")
      return

    # If any two sides added together are less than or equal to the third side, user is told that it is not a triangle and is prompted to re-enter values
    if a + b <= c or a + c <= b or b + c <= a:
      print("Not a triangle.")
      return

    # If/elif/elif statement to classify triangle type
    if a == b == c:
      print("Equilateral")

    elif a == b or a == c or b == c:
      print("Isosceles")

    elif a != b and a != c and b != c:
      print("Scalene")

  except ValueError:
    print("Please enter a float only.")

triangle_check() # Calls the function

```


## 3) Library access hours with membership override

**Prompt**  
Ask for **hour** (int, 0–23) and **member** (`"y"`/`"n"`).  
- Open to **everyone** from 9–20 inclusive.  
- Open to **members only** at hours 7–8 and 21–22.  
- Otherwise, **Closed**.

**Your task**  
- Validate inputs.  
- Print **exactly one** of: `"Open to all"`, `"Open to members"`, `"Closed"`.

*Gentle hint:* Think of this as a small decision tree. Check the “open to all” window first.


```python
def library(): # Defines the function

  # Try/ except to validate the input
  try:
    hour = int(input("Enter a number between 0 and 23: "))
    member = input("Are you a member? y/n: ").strip().lower()

  except ValueError:
    print("Please enter an integer between 0 and 23.")

  # If/elif/else statement to determine which message to print
  if hour >= 9 and hour <= 20:
    print("Open to all.")

  # Nested if/else in an elif to determine which message to print
  elif hour == 7 or hour == 8 or hour == 21 or hour == 22:
    if member == "y":
      print("Open to members.")

    else:
      print("Closed.")

  else:
    print("Closed.")

library() # Calls the function
```


## 4) Temperature banding

**Prompt**  
Ask for **temp** (int) and categorize into exactly one label:  
- `temp <= 0` → `"Freezing"`  
- `1–15` → `"Cold"`  
- `16–25` → `"Mild"`  
- `26–32` → `"Warm"`  
- `>= 33` → `"Hot"`

**Your task**  
- Validate integer input.  
- Make sure **only one** label is printed.

*Gentle hint:* Use `elif` to avoid overlapping outputs.


```python

def temp_value(): # Defines the function
  # Try/except to validate user input
  try:
    temp = int(input("Please enter a temperature: "))

  except ValueError:
    print("Please enter an integer only.")

  # If/elif/else to determine which message to print
  if temp <= 0:
    print("Freezing.")

  elif temp >= 1 and temp <= 15:
    print("Cold.")

  elif temp >= 16 and temp <= 25:
    print("Mild.")

  elif temp >= 26 and temp <= 32:
    print("Warm.")

  else:
    print("Hot")

temp_value() # Calls the function
```
